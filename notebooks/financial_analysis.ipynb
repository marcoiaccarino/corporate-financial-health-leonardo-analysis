```python
def reclassify(ce, sp):
    # Conto Economico a Valore Aggiunto
    vp = ce.loc["Ricavi delle Vendite"]
    ce_cost = ce.loc["Costi per Acquisti e Servizi"]
    va = vp - ce_cost
    l = ce.loc["Costo del Personale"]
    mol = va - l
    aa = ce.loc["Ammortamenti e Svalutazioni"]
    rogc = mol - aa
    rgca = pd.Series([0, 0, 0, 0], index=years)
    roa = rogc + rgca
    of = ce.loc["Proventi / (Oneri) Finanziari"]
    rc = roa + of
    cs = pd.Series([0, 0, 0, 0], index=years)
    rai = rc + cs
    i = ce.loc["Imposte sul Reddito"]
    rn = ce.loc["Utile Netto d'Esercizio"]

    ce_rec = pd.DataFrame(
        {
            "VP (Valore della Produzione)": vp,
            "- CE (Costi Esterni di Gestione)": -ce_cost,
            "= VA (VALORE AGGIUNTO)": va,
            "- L (Costo del Personale / Lavoro)": -l,
            "= MOL (MARGINE OPERATIVO LORDO / EBITDA)": mol,
            "- AA (Ammortamenti e Accantonamenti)": -aa,
            "= ROGC (Redd. Operativo Gestione Caratteristica)": rogc,
            "+ RGCA (Gestione Complementare/Accessoria)": rgca,
            "= ROA (REDDITO OPERATIVO AZIENDALE / EBIT)": roa,
            "- OF (Oneri Finanziari Netti)": of,
            "= RC (Reddito di Competenza / Ordinario)": rc,
            "+/- CS (Componenti Straordinari)": cs,
            "= RAI (Reddito Ante Imposte / EBT)": rai,
            "- I (Imposte sul Reddito)": -i,
            "= RN (REDDITO NETTO / UTILE D'ESERCIZIO)": rn,
        }
    ).T

    # Stato Patrimoniale Finanziario
    ab = (
        sp.loc["Rimanenze di Magazzino"]
        + sp.loc["Crediti Commerciali"]
        + sp.loc["Disponibilità Liquide (Cassa)"]
        + sp.loc["Altre Attività Correnti"]
    )
    afn = sp.loc["Attività Non Correnti (Immobilizzazioni)"]
    ci = ab + afn

    pb = (
        sp.loc["Passività Finanziarie a Breve Termine"]
        + sp.loc["Debiti Commerciali e Altre Passività Correnti"]
    )
    pml = (
        sp.loc["Passività Finanziarie a Lungo Termine"]
        + sp.loc["Fondi Rischi e Oneri"]
    )
    mt = pb + pml
    mp = sp.loc["Patrimonio Netto"]
    ff = mt + mp

    sp_rec = pd.DataFrame(
        {
            "Attivo a Breve (AB)": ab,
            "Attivo Fisso Netto (AFN)": afn,
            "= TOTALE CAPITALE INVESTITO (CI)": ci,
            "Passivo a Breve (PB)": pb,
            "Passivo a Medio-Lungo Termine (PML)": pml,
            "Mezzi di Terzi (MT = PB + PML)": mt,
            "Mezzi Propri (MP)": mp,
            "= TOTALE FONTI DI FINANZIAMENTO (FF)": ff,
        }
    ).T

    return ce_rec, sp_rec


ce_rec_leo, sp_rec_leo = reclassify(ce_leo, sp_leo)
ce_rec_airbus, sp_rec_airbus = reclassify(ce_airbus, sp_airbus)

# Output Leonardo
print(
    "==================== LEONARDO S.p.A. - RICLASSIFICAZIONI (€M) ===================="
)
print("\n--- CONTO ECONOMICO A VALORE AGGIUNTO ---")
display(ce_rec_leo)
print("\n--- STATO PATRIMONIALE FINANZIARIO ---")
display(sp_rec_leo)

# Output Airbus
print(
    "\n\n==================== AIRBUS SE - RICLASSIFICAZIONI (€M) ===================="
)
print("\n--- CONTO ECONOMICO A VALORE AGGIUNTO ---")
display(ce_rec_airbus)
print("\n--- STATO PATRIMONIALE FINANZIARIO ---")
display(sp_rec_airbus)
```

```python
def compute_ratios(ce, sp, ce_rec, sp_rec):
    rn = ce_rec.loc["= RN (REDDITO NETTO / UTILE D'ESERCIZIO)"]
    mp = sp_rec.loc["Mezzi Propri (MP)"]
    rogc = ce_rec.loc["= ROGC (Redd. Operativo Gestione Caratteristica)"]
    ebit = ce_rec.loc["= ROA (REDDITO OPERATIVO AZIENDALE / EBIT)"]
    ci = sp_rec.loc["= TOTALE CAPITALE INVESTITO (CI)"]
    ricavi = ce.loc["Ricavi delle Vendite"]
    mt = sp_rec.loc["Mezzi di Terzi (MT = PB + PML)"]
    df = (
        sp.loc["Passività Finanziarie a Lungo Termine"]
        + sp.loc["Passività Finanziarie a Breve Termine"]
    )
    pml = sp_rec.loc["Passivo a Medio-Lungo Termine (PML)"]
    afn = sp_rec.loc["Attivo Fisso Netto (AFN)"]
    ebitda = ce_rec.loc["= MOL (MARGINE OPERATIVO LORDO / EBITDA)"]
    liq = sp.loc["Disponibilità Liquide (Cassa)"]
    pfn = df - liq
    pb = sp_rec.loc["Passivo a Breve (PB)"]
    ab = sp_rec.loc["Attivo a Breve (AB)"]
    crediti = sp.loc["Crediti Commerciali"] + sp.loc["Altre Attività Correnti"]
    ccn = ab - pb

    ratios = pd.DataFrame(
        {
            "ROE (%) (RN / MP)": (rn / mp) * 100,
            "ROI (%) (ROGC / CI)": (rogc / ci) * 100,
            "ROA (%) (EBIT / CI)": (ebit / ci) * 100,
            "ROS (%) (EBIT / Ricavi)": (ebit / ricavi) * 100,
            "Rotazione Capitale Investito (Ricavi / CI)": ricavi / ci,
            "Rapp. Indebitamento Complessivo (MT / MP)": mt / mp,
            "Rapp. Indebitamento Finanziario (DF / MP)": df / mp,
            "Grado Copertura Attivo Fisso 2° Liv. ((MP+PML)/AFN)": (mp + pml)
            / afn,
            "Incidenza PFN / EBITDA (Anni)": pfn / ebitda,
            "Indice Liquidità Primaria (Cash Ratio) (Liq.Imm / PB)": liq / pb,
            "Indice di Tesoreria (Quick Ratio) ((Liq.Imm+Diff) / PB)": (
                liq + crediti
            )
            / pb,
            "Indice Liquidità Secondaria (Current Ratio) (AB / PB)": ab / pb,
            "Capitale Circolante Netto (€M) (AB - PB)": ccn,
        }
    ).T

    return ratios


ratios_leo = compute_ratios(ce_leo, sp_leo, ce_rec_leo, sp_rec_leo)
ratios_airbus = compute_ratios(
    ce_airbus, sp_airbus, ce_rec_airbus, sp_rec_airbus
)

formatted_ratios_leo = ratios_leo.apply(
    lambda row: (
        [f"{x:.2f}" for x in row]
        if row.name != "Capitale Circolante Netto (€M) (AB - PB)"
        else [f"{x:.2f}" for x in row]
    ),
    axis=1,
)
formatted_ratios_airbus = ratios_airbus.apply(
    lambda row: (
        [f"{x:.2f}" for x in row]
        if row.name != "Capitale Circolante Netto (€M) (AB - PB)"
        else [f"{x:.2f}" for x in row]
    ),
    axis=1,
)

print(
    "==================== INDICI DI BILANCIO: LEONARDO S.p.A. ===================="
)
display(formatted_ratios_leo)

print(
    "\n\n==================== INDICI DI BILANCIO: AIRBUS SE ===================="
)
display(formatted_ratios_airbus)
```

```python
mkt_cap_leo = pd.Series([3600, 4700, 8500, 14800], index=years)
mkt_cap_airbus = pd.Series([88000, 87600, 109000, 125000], index=years)


def compute_altman(ce, sp, sp_rec, mkt_cap):
    ta = sp.loc["Totale Attività"]
    ab = sp_rec.loc["Attivo a Breve (AB)"]
    pb = sp_rec.loc["Passivo a Breve (PB)"]
    ccn = ab - pb
    re = sp.loc["Patrimonio Netto"] * 0.55
    ebit = ce.loc["Risultato Operativo (EBIT)"]
    tl = sp_rec.loc["Mezzi di Terzi (MT = PB + PML)"]
    sales = ce.loc["Ricavi delle Vendite"]

    x1 = ccn / ta
    x2 = re / ta
    x3 = ebit / ta
    x4 = mkt_cap / tl
    x5 = sales / ta

    z_score = 1.2 * x1 + 1.4 * x2 + 3.3 * x3 + 0.6 * x4 + 0.999 * x5

    df_z = pd.DataFrame(
        {
            "X1 (WC/TA)": x1,
            "X2 (RE/TA)": x2,
            "X3 (EBIT/TA)": x3,
            "X4 (MktVal/TL)": x4,
            "X5 (Sales/TA)": x5,
            "Altman Z-Score": z_score,
        }
    ).T

    zones = [
        "Distress Zone 🔴"
        if z < 1.81
        else ("Grey Zone 🟡" if z <= 2.99 else "Safe Zone 🟢")
        for z in z_score
    ]
    df_formatted = df_z.applymap(lambda v: f"{v:.2f}")
    df_formatted.loc["Rischio"] = zones
    return df_formatted


z_leo = compute_altman(ce_leo, sp_leo, sp_rec_leo, mkt_cap_leo)
z_airbus = compute_altman(
    ce_airbus, sp_airbus, sp_rec_airbus, mkt_cap_airbus
)

print(
    "==================== ALTMAN Z-SCORE: LEONARDO S.p.A. ===================="
)
display(z_leo)

print(
    "\n\n==================== ALTMAN Z-SCORE: AIRBUS SE ===================="
)
display(z_airbus)
```
